# 1) List


**Question 1**

```python
import random

my_list = random.sample(range(1, 101), 10)
print(f"The list: {my_list}")

largest_element = max(my_list)
print(f"Largest element: {largest_element}")

smallest_element = min(my_list)
print(f"Smallest element: {smallest_element}")

sum_of_elements = sum(my_list)
print(f"Sum of elements: {sum_of_elements}")

average_of_elements = sum_of_elements / len(my_list)
print(f"Average of elements: {average_of_elements}")
```

**Question 2**

```python
print(f"Original list before modifications: {my_list}")
insert_value = 50
insert_position = 2

my_list.insert(insert_position, insert_value)
print(f"List after inserting {insert_value} at position {insert_position}: {my_list}")

delete_value = 43
if delete_value in my_list:
    my_list.remove(delete_value)
    print(f"List after deleting {delete_value}: {my_list}")
else:
    print(f"Element {delete_value} not found in the list, no deletion. Current list: {my_list}")
my_list.reverse()
print(f"List after reversing: {my_list}")
```

# 2) Nested List

 **Question 1**

```python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
print(f"The 3x3 matrix is:\n{matrix[0]}\n{matrix[1]}\n{matrix[2]}")

second_row = matrix[1]
print(f"Second row: {second_row}")

third_column = [row[2] for row in matrix]
print(f"Third column: {third_column}")

sum_of_elements = 0
for row in matrix:
    sum_of_elements += sum(row)
print(f"Sum of all elements: {sum_of_elements}")
```

**Question 2**

```python
student_marks = [
    {"name": "Alice", "marks": [85, 90, 88]},
    {"name": "Bob", "marks": [78, 82, 80]},
    {"name": "Charlie", "marks": [92, 89, 95]},
    {"name": "David", "marks": [70, 75, 72]}
]

highest_average = -1
student_with_highest_average = ""

print("Student Average Marks:")
for student in student_marks:
    name = student["name"]
    marks = student["marks"]
    average_marks = sum(marks) / len(marks)
    print(f"{name}: {average_marks:.2f}")

    if average_marks > highest_average:
        highest_average = average_marks
        student_with_highest_average = name

print(f"\nStudent with the highest average marks: {student_with_highest_average} ({highest_average:.2f})")
```

# 3)Dictionary

**Question 1**

```python
student_details = {
    "roll_number": "101",
    "department": "Computer Science",
    "cgpa": 3.8
}

print(" Student Details ")
for key, value in student_details.items():
    print(f"{key.replace('_', ' ').title()}: {value}")

new_cgpa = 3.95
student_details["cgpa"] = new_cgpa

print(f"\n After CGPA Update ")
print(f"Updated CGPA for {student_details['roll_number']}: {student_details['cgpa']}")

print("\n--- All Details After Update ---")
for key, value in student_details.items():
    print(f"{key.replace('_', ' ').title()}: {value}")
```

**Question 2**

```python
subject_marks = {
    "Math": 95,
    "Science": 88,
    "English": 92,
    "History": 85,
    "Computer Science": 98
}
print(" Subject Marks ")
print("Subjects (Keys):")
for subject in subject_marks.keys():
    print(f"- {subject}")

print("\nMarks (Values):")
for marks in subject_marks.values():
    print(f"- {marks}")

highest_marks = 0
subject_with_highest_marks = ""

for subject, marks in subject_marks.items():
    if marks > highest_marks:
        highest_marks = marks
        subject_with_highest_marks = subject

print(f"\nSubject with the highest marks: {subject_with_highest_marks} ({highest_marks})")
```

# 4)List in a dictionary

**Question 1**

```python
departments = {
    "HR": ["Alice", "Bob", "Charlie"],
    "Engineering": ["David", "Eve", "Frank", "Grace"],
    "Sales": ["Heidi", "Ivan"],
    "Marketing": ["Judy", "Kevin", "Liam"]
}
print("Department and Employee Details ")
given_department = "Engineering"
if given_department in departments:
    print(f"\nEmployees in {given_department} department: {', '.join(departments[given_department])}")
else:
    print(f"\nDepartment '{given_department}' not found.")
print("\n--- Employee Count per Department ---")
for dept, employees in departments.items():
    print(f"{dept}: {len(employees)} employees")
```

# 5) Dictionary in a List

**Question 1**

```python
students_data = [
    {"name": "Alice", "marks": [85, 90, 88, 92]},
    {"name": "Bob", "marks": [78, 82, 80, 79]},
    {"name": "Charlie", "marks": [92, 89, 95, 91]},
    {"name": "David", "marks": [70, 75, 72, 73]},
    {"name": "Eve", "marks": [90, 93, 89, 96]}
]

print("--- Student Names ---")
for student in students_data:
    print(f"- {student['name']}")

highest_average_marks = -1
topper_name = ""

print("\n--- Calculating Average Marks ---")
for student in students_data:
    name = student["name"]
    marks = student["marks"]
    average_marks = sum(marks) / len(marks)
    print(f"{name}: Average Marks = {average_marks:.2f}")

    if average_marks > highest_average_marks:
        highest_average_marks = average_marks
        topper_name = name

print(f"\n--- Topper of the Class ---")
print(f"The topper is: {topper_name} with an average of {highest_average_marks:.2f} marks.")
```

**Question 2**

```python
books = [
    {"title": "The Hitchhiker's Guide to the Galaxy", "price": 12.99},
    {"title": "Pride and Prejudice", "price": 8.50},
    {"title": "1984", "price": 10.00},
    {"title": "To Kill a Mockingbird", "price": 14.75},
    {"title": "The Great Gatsby", "price": 9.25}
]

print(" Book Inventory ")
for book in books:
    print(f"Title: {book['title']}, Price: ${book['price']:.2f}")

most_expensive_book_title = ""
max_price = 0.0

for book in books:
    if book['price'] > max_price:
        max_price = book['price']
        most_expensive_book_title = book['title']

print(f"\n Most Expensive Book")
print(f"The most expensive book is: '{most_expensive_book_title}' with a price of ${max_price:.2f}")

total_cost = sum(book['price'] for book in books)

print(f"\nTotal Inventory Cost ")
print(f"The total cost of all books is: ${total_cost:.2f}")
```

# 6) Array

**Question 1**

```python
import array as arr
import random

my_array = arr.array('i', [random.randint(1, 100) for _ in range(10)])

print(f"The array: {list(my_array)}")

maximum_element = max(my_array)
print(f"Maximum element: {maximum_element}")

minimum_element = min(my_array)
print(f"Minimum element: {minimum_element}")

sum_of_elements = sum(my_array)

average_of_elements = sum_of_elements / len(my_array)
print(f"Average of elements: {average_of_elements:.2f}")
```

**Question 2**

```python
print(f"Original array before modifications: {list(my_array)}")

insert_value = 55
insert_position = 3
my_array.insert(insert_position, insert_value)
print(f"Array after inserting {insert_value} at position {insert_position}: {list(my_array)}")

delete_value = my_array[0]
if delete_value in my_array:
    my_array.remove(delete_value)
    print(f"Array after deleting {delete_value}: {list(my_array)}")
else:
    print(f"Element {delete_value} not found in the array, no deletion performed. Current array: {list(my_array)}")

my_array.reverse()
print(f"Array after reversing: {list(my_array)}")
```

# 8) Set

**Question 1**



```python
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}

print(f"Set 1: {set1}")
print(f"Set 2: {set2}")

union_set = set1.union(set2)
print(f"Union of Set 1 and Set 2: {union_set}")

intersection_set = set1.intersection(set2)
print(f"Intersection of Set 1 and Set 2: {intersection_set}")

difference_set1_2 = set1.difference(set2)
print(f"Difference (Set1 - Set2): {difference_set1_2}")

difference_set2_1 = set2.difference(set1)
print(f"Difference (Set2 - Set1): {difference_set2_1}")

symmetric_difference_set = set1.symmetric_difference(set2)
print(f"Symmetric Difference: {symmetric_difference_set}")
```
