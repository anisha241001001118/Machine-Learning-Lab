```python
import pandas as pd

#Series
s = pd.Series(range(1, 11))

#Series into DataFrame
df = s.to_frame(name="Numbers")

print("Series:")
print(s)

print("\nDataFrame:")
print(df)

print("\nShape:", df.shape)
print("Data type:", s.dtype)
print("Sum:", s.sum())
print("Mean:", s.mean())
print("Maximum:", s.max())
print("Minimum:", s.min())
```

# Question 2

```python
import pandas as pd
rows = ["R1", "R2", "R3"]
columns = ["C1", "C2", "C3"]
A = pd.DataFrame(
    [[1, 2, 3],
     [4, 5, 6],
     [7, 8, 9]],
    index=rows,
    columns=columns
)
B = pd.DataFrame(
    [[9, 8, 7],
     [6, 5, 4],
     [3, 2, 1]],
    index=rows,
    columns=columns
)

print("DataFrame A:")
print(A)
print("\nDataFrame B:")
print(B)

print("\nAddition:")
print(A + B)
print("\nSubtraction:")
print(A - B)
print("\nElement-wise Multiplication:")
print(A * B)
print("\nMatrix Multiplication:")
print(A.values.dot(B.values))
```

# Question 3

```python
import pandas as pd

data = {
    "Student": ["Amit", "Bhavna", "Chirag", "Divya", "Esha"],
    "Maths": [85, 70, 95, 60, 88],
    "Science": [90, 65, 91, 75, 84],
    "English": [88, 72, 93, 70, 86],
    "History": [92, 68, 96, 80, 90]
}
df = pd.DataFrame(data)

df.set_index("Student", inplace=True)
df["Total"] = df.sum(axis=1)

df["Average"] = df[["Maths", "Science", "English", "History"]].mean(axis=1)

print("Student Data:")
print(df)

print("\nAverage marks of each student:")
print(df["Average"])

print("\nAverage marks of each subject:")
print(df[["Maths", "Science", "English", "History"]].mean())

print("\nHighest mark:")
print(df[["Maths", "Science", "English", "History"]].max().max())

print("\nLowest mark:")
print(df[["Maths", "Science", "English", "History"]].min().min())

print("\nStudent with highest total marks:")
print(df["Total"].idxmax())

print("\nSorted by Total Marks:")
print(df.sort_values(by="Total", ascending=False))
```

# Question 4

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Day": ["Day1", "Day2", "Day3", "Day4",
            "Day5", "Day6", "Day7"],
    "Temperature": [30, np.nan, 32, 29, np.nan, 35, 31]
})

print("Original DataFrame:")
print(df)

print("\nMissing values:")
print(df.isnull())

mean_temp = df["Temperature"].mean()
print("\nMean Temperature:", mean_temp)

df_mean = df.copy()
df_mean["Temperature"] = df_mean["Temperature"].fillna(mean_temp)

print("\nAfter filling with mean:")
print(df_mean)

df_ffill = df.copy()
df_ffill["Temperature"] = df_ffill["Temperature"].ffill()

print("\nAfter Forward Fill:")
print(df_ffill)

df_drop = df.dropna()

print("\nAfter dropping missing rows:")
print(df_drop)

print("\nAverage temperature after cleaning:",
      df_drop["Temperature"].mean())
```

# Question 5

```python
import pandas as pd

data = {
    "Product": ["Product A", "Product B", "Product C", "Product D"],
    "Jan": [120, 100, 150, 90],
    "Feb": [130, 110, 145, 100],
    "Mar": [140, 120, 160, 110],
    "Apr": [150, 130, 170, 120],
    "May": [160, 140, 180, 130],
    "Jun": [170, 150, 190, 140]
}
df = pd.DataFrame(data)

df.set_index("Product", inplace=True)
print("Sales Data:")
print(df)

total_sales = df.sum(axis=1)
print("\nTotal sales of each product:")
print(total_sales)

average_sales = df.mean(axis=1)
print("\nAverage monthly sales:")
print(average_sales)

best_product = total_sales.idxmax()
print("\nBest-selling product:")
print(best_product)

monthly_total = df.sum(axis=0)
best_month = monthly_total.idxmax()
print("\nMonth with highest overall sales:")
print(best_month)

std_sales = df.std(axis=1)
print("\nStandard deviation:")
print(std_sales)

df["Q1"] = df[["Jan", "Feb", "Mar"]].sum(axis=1)
df["Q2"] = df[["Apr", "May", "Jun"]].sum(axis=1)
pivot_table = df[["Q1", "Q2"]]

print("\nQuarterly Sales Pivot Table:")
print(pivot_table)
```
